```python
# ==========================================================
# SMARTVISION AI - COMPLETE EDA
# ==========================================================

import os
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

sns.set_style("whitegrid")
```

```python

# ==========================================================
# DATASET PATH
# ==========================================================

BASE_PATH = r"C:\Users\sksaa\OneDrive\Desktop\smartVision AI\smartvision_dataset\classification"

TRAIN_PATH = os.path.join(BASE_PATH, "train")
VAL_PATH = os.path.join(BASE_PATH, "val")
TEST_PATH = os.path.join(BASE_PATH, "test")
```

```python

# ==========================================================
# CLASS DISTRIBUTION
# ==========================================================

class_counts = {}

for cls in os.listdir(TRAIN_PATH):

    class_path = os.path.join(TRAIN_PATH, cls)

    if os.path.isdir(class_path):

        count = len([
            f for f in os.listdir(class_path)
            if f.lower().endswith((".jpg", ".jpeg", ".png"))
        ])

        class_counts[cls] = count

df_classes = pd.DataFrame({
    "Class": list(class_counts.keys()),
    "Count": list(class_counts.values())
})

print("\nCLASS DISTRIBUTION")
print(df_classes)
```

```python

# ==========================================================
# BAR CHART
# ==========================================================

plt.figure(figsize=(15,6))
sns.barplot(
    data=df_classes,
    x="Class",
    y="Count"
)

plt.xticks(rotation=90)
plt.title("Class Distribution")
plt.show()
```

```python

# ==========================================================
# PIE CHART
# ==========================================================

plt.figure(figsize=(10,10))

plt.pie(
    df_classes["Count"],
    labels=df_classes["Class"],
    autopct="%1.1f%%"
)

plt.title("Class Distribution Pie Chart")
plt.show()
```

```python

# ==========================================================
# SAMPLE IMAGES
# ==========================================================

plt.figure(figsize=(15,12))

for i, cls in enumerate(list(class_counts.keys())[:9]):

    class_folder = os.path.join(TRAIN_PATH, cls)

    image_files = [
        f for f in os.listdir(class_folder)
        if f.lower().endswith((".jpg",".jpeg",".png"))
    ]

    if len(image_files) > 0:

        img_path = os.path.join(class_folder, image_files[0])

        img = Image.open(img_path)

        plt.subplot(3,3,i+1)
        plt.imshow(img)
        plt.title(cls)
        plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import os
from PIL import Image
import numpy as np

train_path = r"C:\Users\sksaa\OneDrive\Desktop\smartVision AI\smartvision_dataset\classification\train"

worst = []

for cls in os.listdir(train_path):

    class_path = os.path.join(train_path, cls)

    if os.path.isdir(class_path):

        for img_name in os.listdir(class_path):

            try:
                img_path = os.path.join(class_path, img_name)

                img = Image.open(img_path).convert("RGB")
                img = np.array(img)

                black_pixels = np.sum(np.all(img == [0,0,0], axis=-1))
                total_pixels = img.shape[0] * img.shape[1]

                pct = (black_pixels / total_pixels) * 100

                if pct > 80:
                    worst.append((pct, img_path))

            except:
                pass

worst = sorted(worst, reverse=True)

print("Images with >80% black area:", len(worst))

for pct, path in worst[:20]:
    print(f"{pct:.2f}% -> {path}")
```

```python

# ==========================================================
# IMAGE SIZE ANALYSIS
# ==========================================================

widths = []
heights = []

for cls in os.listdir(TRAIN_PATH):

    class_folder = os.path.join(TRAIN_PATH, cls)

    if os.path.isdir(class_folder):

        image_files = [
            f for f in os.listdir(class_folder)
            if f.lower().endswith((".jpg",".jpeg",".png"))
        ]

        for img_name in image_files:

            try:

                img = Image.open(
                    os.path.join(class_folder, img_name)
                )

                w, h = img.size

                widths.append(w)
                heights.append(h)

            except:
                pass

print("\nAverage Width :", round(np.mean(widths),2))
print("Average Height:", round(np.mean(heights),2))
print("Minimum Width :", min(widths))
print("Maximum Width :", max(widths))
print("Minimum Height:", min(heights))
print("Maximum Height:", max(heights))
```

```python

# ==========================================================
# WIDTH DISTRIBUTION
# ==========================================================

plt.figure(figsize=(10,5))

sns.histplot(
    widths,
    bins=30,
    kde=True
)

plt.title("Image Width Distribution")
plt.show()
```

```python

# ==========================================================
# HEIGHT DISTRIBUTION
# ==========================================================

plt.figure(figsize=(10,5))

sns.histplot(
    heights,
    bins=30,
    kde=True
)

plt.title("Image Height Distribution")
plt.show()
```

```python

# ==========================================================
# RGB PIXEL DISTRIBUTION
# ==========================================================

pixels = []

for cls in list(class_counts.keys())[:5]:

    class_folder = os.path.join(TRAIN_PATH, cls)

    image_files = [
        f for f in os.listdir(class_folder)
        if f.lower().endswith((".jpg",".jpeg",".png"))
    ][:10]

    for img_name in image_files:

        img = cv2.imread(
            os.path.join(class_folder, img_name)
        )

        if img is not None:

            pixels.extend(
                img.reshape(-1,3)
            )

pixels = np.array(pixels)

plt.figure(figsize=(12,6))

plt.hist(
    pixels[:,0],
    bins=50,
    alpha=0.5,
    label="Blue"
)

plt.hist(
    pixels[:,1],
    bins=50,
    alpha=0.5,
    label="Green"
)

plt.hist(
    pixels[:,2],
    bins=50,
    alpha=0.5,
    label="Red"
)

plt.legend()
plt.title("RGB Pixel Distribution")
plt.show()
```

```python

# ==========================================================
# TRAIN / VAL / TEST DISTRIBUTION
# ==========================================================

train_count = sum([
    len(files)
    for _, _, files in os.walk(TRAIN_PATH)
])

val_count = sum([
    len(files)
    for _, _, files in os.walk(VAL_PATH)
])

test_count = sum([
    len(files)
    for _, _, files in os.walk(TEST_PATH)
])

split_df = pd.DataFrame({
    "Split":["Train","Validation","Test"],
    "Images":[train_count,val_count,test_count]
})

print("\nDATASET SPLITS")
print(split_df)

plt.figure(figsize=(8,5))

sns.barplot(
    data=split_df,
    x="Split",
    y="Images"
)

plt.title("Train-Val-Test Distribution")
plt.show()
```

```python

# ==========================================================
# DATASET SUMMARY
# ==========================================================

summary = pd.DataFrame({

    "Metric":[
        "Total Classes",
        "Total Train Images",
        "Total Validation Images",
        "Total Test Images",
        "Average Width",
        "Average Height"
    ],

    "Value":[
        len(class_counts),
        train_count,
        val_count,
        test_count,
        round(np.mean(widths),2),
        round(np.mean(heights),2)
    ]
})

print("\nFINAL DATASET SUMMARY")
print(summary)
```
